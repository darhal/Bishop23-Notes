## The rules of probability
### The sum and product rule

$p(X=x_i)$ is the probability that the random variable X can take the value of $x_i$. We can simple write $p(X)$ to denote a distrubtion over the random variable $X$ and $p(x_i)$ to denote a distrubtion evaluated for the particular value $x_i$.

We can write the two fundamental rules of probability theory as follows :
**Sum rule :**
$$p(X) = \sum_{Y}{} p(X,Y)$$

**Product rule :**
$$p(X,Y) = p(X|Y)p(X)$$

### Bayes Theorem
From the afromentioned product rule and symmetry property $p(X,Y)=p(Y,X)$ we can obtain : 
$$p(Y|X) = \frac{p(X|Y)p(Y)}{p(X)}$$

The Bayes theorem relates the conditional distrubtion $p(Y|X)$ to the reversed conditional distrubtion $p(X|Y)$. Using the sum rule the denominator in Baes theorem can be :
$$P(X) = \sum_{Y} p(X|Y)p(Y)$$
Hence, we can see the deniminator in Bayes theorem as being the normalization constant required to ensure that the sum over the conditional probability distrubtion $P(Y|X)$ overall values of $Y$ equals one : 
$$p(Y|X) = \frac{p(X|Y)p(Y)}{\sum_{Y}{p(X|Y)p(Y)}}$$

### Prior and posterior probabilities
The most relevant probability we can before knowing any extra information on $X$ is $P(X)$ which is called the *prior probability* because its the probability available we ovbserve the result of the extra information. We can then use Bayes theorem to compute th e probability $p(X|Y)$ which is called *posterior probability* because its the probability obtained after we observed the result of the extra information.

### Indepandent variables
If the joint distribution of two variables factorizes into the product of the
marginals, so that $p(X,Y) = p(X)p(Y)$, then $X$ and $Y$ are said to be independent. An example of independent events would be the successive flips of a coin. From the product rule, we see that $p(Y|X) = p(Y)$, and so the conditional distribution of $Y$ given $X$ is indeed independent of the value of $X$.

## Probability Densisties
We can't simply extend probability concepts discussed so far to continious variables since the probabiliy of observing a specific value for a continuous variable to infinite precision will be effectively zero. We need to introduce the concept of probability density. 

We define the probability density $p(x)$ over a continuous variable $x$ to be such that the probability of $x$ falling in the interval $(x, x+δx)$ is given by p(x)δx for $δx →0$.
$$p(x \in (a,b)) = \int_{a}^{b}p(x)dx$$

the probability density $p(x)$ must satisfy two conditions : 
$$p(X) \geq 0$$
$$\int_{-\infty}^{+\infty}p(x)dx = 1$$

The probability that $x$ lies in $(-\infnity, z)$$ is given by the *cumlative distrubtion function* : 
$$P(Z) = \int_{-\infty}^{z} p(x)dx$$

Which satisfies $P'(x) = p(x)$ the density is the derivative of the CDF.

**Sum rule** : $$p(x) = \int_{}p(x,y)dy$$

**Product rule** : $$p(x,y)=p(y|x)p(x)$$

**Bayes theorem** : $$p(y|x)=\frac{p(x|y)p(y)}{p(x)}$$
Where the denominator is given by :
$$p(x)=\int_{}p(x|y)p(y)dy$$

### Example distrubtions 
$p(x)$ is a constant independent of x is an example of a distrubtion however this cannot be normalized because the integral will be divergent. Distruvtions that cannot be normalized are called *improper*. We can have the unifrom distrubtion that is constant over a finite region $(c,d)$ and zero elsewhere :
$$p(x)=\frac{1}{d-c}, x \in (c,d)$$

Another form of density is the exponential distrubtion given by :
$$p(x|\lambda)=\lambda exp(-\lambda x), x \geq 0$$

The *Laplace dsitrbution* is a variant of the exponential distrbution and it allows the peak to be moved to a location $\mu$ :
$$p(x|\mu,\gamma)=\frac{1}{2\gamma}exp(-\frac{|{x-\mu}|}{\gamma})$$

*Dirac delta function* is defined to be 0 everywhere except at $x=\mu$ written as :
$$p(x|\mu)=\delta(x-\mu)$$

For a finite set of observations of $x$ given by $D=\{x_1, ..., x_N\}$ then we can construct an *empirical distrubtion* using the delta function which consists of a Dirac delta function centered at each of the data points :
$$p(x|D)=\frac{1}{N}\sum_{n=1}^{N}\delta(x-x_n)$$

### Expectation and co-variance 
The weighted average of a function $f(x)$ under a probability distrubtion $p(x)$ is called the *expectation of* $f(x)$ and is denoted by 
$$\mathbb{E}[f] = \sum_{x}p(x)f(x)$$
For continious variables :
$$\mathbb{E}[f] = \int_{}p(x)f(x)dx$$
In either case, given a finite number $N$ of points drawn from the probability distrubtion/density then the expectation can be approximated as : 
$$\mathbb{E}[f] = \frac{1}{N}\sum_{x=1}^{N}f(x_n)$$
This approx becomes exact when $N\rightarrow\infty$. This approximation is called *Monte Carlo* approximation. When the samples $x_n$ are drawn from the distribution $p(x)$, the weighting is implicitly handled by the sampling process as we sample more frequently from areas where $p(x)$ is high.

We can calculate the expectation of functions of several variables, we use a subscript to indicate which variable is being averaged over : $\mathbb{E}_x[f(x,y)]$ denotes the average of $f(x,y)$ w.r.t. distrubtion of $x$.

We can also consider *conditional expectation* w.r.t. conditional distrubtion : $\mathbb{E}_x[f|y]=\sum_{x}p(x|y)f(x)$ for vcontinious var. we replace sum by integral.

The *variance* of $f(x)$ provides the measure of how much $f(x)$ varies arround its mean value : 
$$var[f]=\mathbb{E}[(f(x)-\mathbb{E}[f(x)])^2]$$
or 
$$var[f]=\mathbb{E}[x^2]-{\mathbb{E}[x]}^2$$

For 2 random vars $x$ and $y$, the covariance is the extent to which the 2 variables vary togther :
$$cov[x,y]=\mathbb{E_{x,y}}[{x-\mathbb{E}[x]}{y-\mathbb{E}[y]}] = \mathbb{E_{x,y}[xy]} - \mathbb{E}[x]\mathbb{E}[y]$$

If $x$ and $y$ are independent then their covariance is 0. For 2 vectors $x$ and $y$ their cov matrix is given by :
$$cov[x,y]=\mathbb{E_{x,y}}[\{x-\mathbb{E}[x]\}\{y^T-\mathbb{E}[y^T]\}]=\mathbb{E_{x,y}}[xy^T]-\mathbb{E}[x]\mathbb{E}[y^T]$$

The covariance of thge comonent of a vector $x$ with each other is noted $cov[x]=cov[x,x]$

## The Gaussian Distribution

Called *normal* or *Gaussian* distribution : 
$$N(x|\mu,\sigma^2)=\frac{1}{(2\pi\sigma^2)^{1/2}}exp\{-\frac{1}{2\sigma^2}(x-\mu)^2\}$$

Where $\mu$ is the *mean* and $\sigma^2$ is the *variance*, $\sigma$ is the *standard deviation* and $\beta=1/\sigma^2$ is the *precision*.

### Mean and variance
Because \mu represents the average value of $x$ under the distrubtion, its referred to as the mean. 
$$\mathbb{E}[x]=\mu$$

The second order moment (expectation of $x^2$) is : 
$$\mathbb{E}[x^2]=\int_{-\infty}^{\infty}N(x|\mu,\sigma^2)x^2dx=\mu^2+\sigma^2$$
Hence the variance of $x$ is :
$$var[x]=\mathbb{E}[x^2]-\mathbb{E}[x]^2=\sigma^2$$
The maximum of a distri is known as the mode. For a Gaussian the mode is the mean.

### Likelihood function
Suppose we have a row vector $x=(x_1,...,x_N)$ representing N observation of the scalar variable $x$ and that the observations are drawn independently from a Gaussian distri whose mean and variance are unknown and we would like to determine these parameters from the data set. The problem of estimating a distri given a finite set of observations is known as *density estimation*. Since there is inifnitely many probability distrubtions that could have given rise to the observed finite data set, the problem of density estimation is fundamentally ill-posed. For ou example we constrain the space of distributions to be Gaussian which leads to a well-defined solution.

Data points that are drawn independently fromt he same distrubtion are said to be indepenedant and identically distrbuted (iid). The joint probability of two independent event is given by the product of the marignal probabilities for each even seperately. Since $x$ is iid we can write the probability of the data set given $\mu$ and $\sigma^2$ : 
$$p(x|\mu,\sigma^2)=\prod_{n=1}^{N}N(x_n|\mu,\sigma^2)$$
When viewed as a function of $\mu$ and $\sigma^2$ this is called the *likelihood function* for the Gaussian distri.

A common approach to find these paramters using an observed data set is to find paramters that maximise the likelihood (called *maximum likelihood*)

To maximise the likelihood function its more convinent to maximize the log of it since the log is monotonically increasing the log this simplifies the math. The log also helps numerically since the product of lartge number of small probabilities can easily underflow the precision (in computers) : 
$$ln(p(x|\mu,\sigma^2))=-\frac{1}{2\sigma^2}\sum_{n=1}^{N}(x_n-\mu)^2-\frac{N}{2}ln(\sigma^2)-\frac{N}{2}ln(2\pi)$$
Maximizing this equation wrt $\mu$ we obtain the *sample mean* (the mean of the observed values $\{x_n\}$):
$$\mu_{ML}=\frac{1}{N}\sum_{n=1}{N}x_n$$
Similarly, maximizing wrt to $\sigma^2$ we onbtain the *sample variance* wrt to the sample mean :
$$\sigma_{ML}^2 = \frac{1}{N}\sum_{n=1}^{N}(x_n-\sigma_{ML})^2$$

### Bias of maximum likelihood
The maxi likelihood have some limitations. The maxi likelihood solutions $\mu_{ML}$ and $\sigma_{ML}^2$ are functions of dataset values $x_1, ..., x_n$ generated independently from a Gaussian $N(\mu, \sigma^2)$ the expectations of $\mu_{ML}$ and $\sigma_{ML}^2$ wrt to these dataset values are : 
$$\mathbb{E}[\mu_{ML}] = \mu$$
$$\mathbb{E}[\sigma_{ML}^2]=(\frac{N-1}{N})\sigma^2$$
When averaged over data sets of a given size, the ML for the mean will equal the true mean but the ML estimate for the variance will underestimate the true variance by $\frac{N-1}{N}$. This is known as the *bias* in which the estimator of a random quantity is systematically different from the true value.

The bias arises because the variance is measured relative to the ML of the mean which itself is tuned to the data. If we had access to the true mean $\mu$ and used it to determine the variance we get : 
$$\hat{\sigma}^2=\frac{1}{N}\sum_{n=1}^{N}(x_n-\mu)^2$$

Then we find that $\mathbb{E}[{\hat{\sigma}^2}]=\sigma^2$ which is unbiased. But we dont have the access to the true mean but only the observed data values. It follows that for a Gaussian distri the following estimate for the variance parameter is unbiased : 
$$\hat{\sigma}^2=\frac{N}{N-1}\sigma_{ML}^2=\frac{1}{N-1}\sum_{n=1}^{N}(x_n-\mu_{ML})^2$$

Correcting for the bias of ML in complex models such as neural networks is not so easy, however.

Note that the bias of ML solutions becomes less significant as the N of data points increase. When $N\rightarrow\infty$ the ML solution for the variance equals the true variance. In case of Gaussian anything other than small N, this bias is not a serious problem. However, with complex models with many paramters for which the bias problems associated with ML will be much more severe, this issue is closely related to the problem of *over-fitting*.
